## Positional-aware LSTM

The code used for running the model is taken from the original code proposed by Yuhao Zhang, one of the author of the paper "**Position-aware Attention and Supervised Data Improve Slot Filling**", available at the following link https://github.com/yuhaozhang/tacred-relation. 

The proposed implementation consists of a 2-layer stacked unidirectional LSTM with dropout and an attention layer, which takes a concatenation of word embeddings, POS and NER embeddings as input. The parameters for training the model are left unmodified, except for the number of training epochs which is reduced to 20. Furthermore, slight variations are apported to the original code for allowing evaluation of the model on both TACRED and TACREV dev set without the need to retrain the model (as we have the same training data in both cases).


```python
train.model_train(
    data_dir="../../../data/",
    vocab_dir="./",
    emb_dim=300,
    ner_dim=30,
    pos_dim=30,
    hidden_dim=200,
    num_layers=2,
    dropout=0.5,
    word_dropout=0.04,
    topn=1e10,
    lower=False,
    attn=True,
    attn_dim=200,
    pe_dim=30,
    lr=1.0,
    optim="sgd",
    num_epoch=20,
    batch_size=50,
    max_grad_norm=5.0,
    log_step=20,
    log="logs.txt",
    save_epoch=15,
    save_dir="./saved_models",
    id="00",
    info="",
    seed=1234,
    cuda=torch.cuda.is_available(),
    cpu=False,
)
```

We obtain two different models. However, we decided to focus only on the one obtained considering TACREV dev set, as it is a revisited version, modified by linguistic experts and we believe that these changes can be beneficial for the performance of the model. For the same reason, the best obtained model, is finally evaluated on TACREV test set. 

```python
# Test the model evaluated on TACREV dev set on TACREV test set and store results in a file
eval.evaluate_saved_model(
    model="/best_model_tacrev.pt",
    data_dir="../../../data/tacrev/json",
    out="./evaluation/errors_TACREV_model.txt",
)
```
